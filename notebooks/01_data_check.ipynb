## shape 요약
### 목적
head(), tail(),shape_summary를 통해 shape를 확인 한다
### 실행 전 예상
shape로는 |데이터셋|예상 행|예상 열|이 나타나고 head()는 상위 데이터셋 5개, tail()은 하위 데이터셋 5개를 보여줄 것으로 예상된다

```python
shape_summary = pd.DataFrame(

    [
        {
            "dataset": name,
            "rows": df.shape[0],
            "columns": df.shape[1],
        }
        for name, df in datasets.items()
    ]
)

print(shape_summary)
```

### 검증
- 각 데이터 셋을 들어가 실제 상위, 하위 데이터, 행과 열의 합을 확인하였다.

- 두 값이 일치했다.
### 결과 해석
각 데이터 셋의 구조와 head, tail을 확인할 수 있다.

## 결측치 요약
### 목적
각 데이터의 결측지를 확인한다
### 실행 전 예상
각 데이터의 결측치 개수, 결측치 비율을 보여줄 것으로 예상된다.

```python
for name, df in datasets.items():

    print(f"\n[{name}]")
    display(df.isna().sum().to_frame("missing_count"))
```

```python
missing_summaries = []

for name, df in datasets.items():
    summary = pd.DataFrame(
        {
            "missing_count": df.isna().sum(),
            "missing_rate_pct": (
                df.isna().mean() * 100
            ).round(2),
        }
    )

    summary.insert(0, "dataset", name)
    summary.insert(1, "column", summary.index)
    missing_summaries.append(summary.reset_index(drop=True))

missing_summary = pd.concat(
    missing_summaries,
    ignore_index=True
)

missing_summary
```

## 실제 결과
각 데이터의 결측치가 없고, 이에 따라 결측치 비율이 0으로 나왔다,
### 검증
- 실제로 각 데이터 파일에 들어가 결측치가 있는지 확인하였다.
- 눈으로 확인한 값으로 결측치 비율을 계산하였다
- 두 값이 일치했다.
### 결과 해석
결측치 개수와 결측치 비율 확인 결과 코딩이 문제 없음을 확인 하였다.

## 날짜 타입 변환
### 목적
str 타입의 날짜를 datetime 타입으로 변환한다.
### 실행 전 예상
date가 있는 str 정보를 datetime으로 고정한다.


```python
from pathlib import Path
import pandas as pd

# 1. 상위 폴더(..)로 올라가서 data/raw 접근
data_dir = Path('..') / 'data' / 'raw'

# 2. CSV 파일 불러오기
customers = pd.read_csv(data_dir / 'customers.csv')
orders = pd.read_csv(data_dir / 'orders.csv')
order_items = pd.read_csv(data_dir / 'order_items.csv')
products = pd.read_csv(data_dir / 'products.csv')

# 3. 날짜 컬럼이 있는 데이터셋만 변환하는 함수
def convert_date_columns(dfs_dict):
    for name, df in dfs_dict.items():
        # 컬럼명에 'date'가 들어간 컬럼만 추출
        date_cols = [col for col in df.columns if 'date' in col.lower()]
        
        # 날짜 컬럼이 있는 경우만 pd.to_datetime 실행
        if date_cols:
            for col in date_cols:
                df[col] = pd.to_datetime(df[col])
                print(f"[{name}] '{col}' 변환 완료 -> {df[col].dtype}")
        else:
            print(f"[{name}] 날짜 컬럼 없음 (변환 생략)")
            
    return dfs_dict

# 4. 함수 실행
dfs = {
    'customers': customers,
    'orders': orders,
    'order_items': order_items,
    'products': products
}

dfs = convert_date_columns(dfs)
```

### 실제 결과
custormers와 orders 파일의 날짜 데이터를 datetime 타입으로 변환하였고 order_items와 products는 날짜 데이터가 없어 변환생략 하였다.
### 검증
- 날짜 컬럼 존재 여부 확인: 전체 4개 데이터셋 중 customers와 orders 2개 테이블에서만 날짜 컬럼 정확히 식별
- 데이터 타입 변경 확인
[customers] 'signup_date' 변환 완료 -> datetime64[us]
[orders] 'order_date' 변환 완료 -> datetime64[us]
- 날짜 데이터가 없는 csv 파일은 생략한 것 확인
### 결과 해석
str 타입의 날짜 데이터를 datetime 타입으로 변환하였다.
